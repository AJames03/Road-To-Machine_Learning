## What is Normalization?
- **Data normalization** is a technique used to organize and scale data so that it becomes easier and more efficient to work with — especially in machine learning, data analysis, and databases.

### In Machine Learning
- **Data normalization** means adjusting the values of features (columns) so that they share a common scale — usually **between 0 and 1** or with a mean of 0 and standard deviation of 1.
- **Purpose:**
    - To ensure that no feature dominates because of its scale.

    - To improve the performance and training speed of machine learning models.

    - To make distance-based algorithms (like KNN or SVM) work better.

## Common Methods:

### 1. Min-Max Normalization:
**Formula:**
$$
X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}
$$

**Example:**
```
Score: [50, 60, 70, 80, 90, 100]
```
**Situation:** 

Let's say the normalize in Score is X = 50.
- X = 50
- Xmin = 50 ~ *This is the Minimum Value in the datasets*
- Xmax = 100 ~ *This is the Maximum Value in the datasets*

$$
X_{norm} = \frac{50 - 50}{100 - 50}
$$

**Solution**
$$
\frac{ 50 - 50}{100 - 50} = \frac{0}{50} = 0
$$

### 2. Z-Score Normalization:

**Formula:**
*Z-Score Formula:*
$$
Z = \frac{X - \mu}{\sigma}
$$
- ***X*** = Score
- ***$\mu$*** = Mean
- ***$\sigma$*** = Standard Deviation

**Example:**
```
Score: [50, 60, 70, 80, 90, 100]
```
**Step 1:**
- Find the Mean
$$
\mu = \frac{50 + 60 + 70 + 80 + 90 + 100}{6} = \frac{450}{6} = 75
$$

**Step 2:**
- Find the Standard Deviation
$$
\sigma = \sqrt{\frac{(50 - 75)^2 + (60 - 75)^2 + (70 - 75)^2 + (80 - 75)^2 + (90 - 75)^2 + (100 - 75)^2}{6}}
$$

$$
\sigma = \sqrt{\frac{625 + 225 + 25 + 25 + 225 + 625}{6}}
$$

$$
\sigma = \sqrt{\frac{1750}{6}} \approx \sqrt{291.67} \approx 17.08
$$

## Conclusion:

### Why it is important?

**1. Ensures Fair Comparison Between Features**
- Features (columns) may have different units and scales (e.g., height in cm vs income in dollars).

- Without normalization, a feature with larger values can dominate the model.

*Normalization puts all features on the same scale — like 0 to 1 or -1 to 1.*

**2. Improves Accuracy and Speed of Algorithms**

Many machine learning algorithms (like KNN, SVM, logistic regression, and neural networks) are sensitive to the scale of data.

*Normalized data helps:*

- Models converge faster

- Avoid bias due to large values

- Improve accuracy


**3. Helps with Distance-Based Algorithms**

Algorithms like:

- K-Nearest Neighbors (KNN)

- K-Means clustering

- Principal Component Analysis (PCA)

*rely on distance (like Euclidean). If features are not normalized, the distance will be distorted.*

**4. Reduces Redundancy and Improves Efficiency (Databases)**

In databases (not ML), normalization organizes data into tables to:

- Reduce duplication

- Improve data integrity
